```python
import os
import glob
from pathlib import Path

current_dir = Path.cwd()
data_dir = current_dir.parent/'data'
```

```python
import requests
from bs4 import BeautifulSoup as bs
from pprint import pprint
```

```python
landingPage_url= 'https://www.elections.il.gov/Downloads/CampaignDisclosure/CampaignDisclosureDataFiles/'

landingPage_resposne = requests.get(landingPage_url)

if landingPage_resposne.status_code == 200:
    data = bs(landingPage_resposne.text, 'html.parser')
    landingPage_files = {}
    for a in data.select('a')[1:]:
        file_name = a.get_text(strip=True)
        file_url = landingPage_url + file_name
        landingPage_files[file_name] = file_url
else:
    print(f'HTTP Error {landingPage_resposne.status_code}: {landingPage_resposne.reason}')
```

```python
from get_illinoisFinance import get_illinoisFinance
from time import sleep
from random import randint
```

```python
for k, v in landingPage_files.items():
    url = v
    destination= f'{data_dir}/{k}'
    get_illinoisFinance(url, destination)
    sleep_timer =randint(5, 20)
    print(f'About to pause for {sleep_timer} seconds.')
    sleep(sleep_timer)
```
